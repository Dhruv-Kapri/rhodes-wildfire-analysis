```python
api_url = "https://earth-search.aws.element84.com/v1"
```

```python
from pystac_client import Client

client = Client.open(api_url)
```

```python
collections = client.get_collections()

for collection in collections:
    print(collection)
```

```python
collection_sentinel_2_l2a = "sentinel-2-l2a"
```

```python
from shapely.geometry import Point
point = Point(27.95, 36.20)  # Coordinates of a point on Rhodes
```

```python
search = client.search(
    collections=[collection_sentinel_2_l2a],
    intersects=point,
)

print(search.matched())
```

```python
search = client.search(
    collections=[collection_sentinel_2_l2a],
    intersects=point,
    datetime="2023-07-01/2023-08-31"
)

print(search.matched())
```

```python
items = search.item_collection()

print(len(items))
```

```python
for item in items:
    print(item)
```

```python
item = items[0]
print(item.datetime)
```

```python
print(item.geometry)
print("-"*20)

for k, v, *_ in item.properties.items():
    print(f"{k}: {v}")
```

```python
search = client.search(
    collections=[collection_sentinel_2_l2a],
    intersects=point,
    datetime="2023-07-01/2023-08-31",
    query={"eo:cloud_cover": {"lt": 0.1}}
)

print(search.matched())
```

```python
items = search.item_collection()
items.save_object("../outputs/rhodes_sentinel-2.json")
```
