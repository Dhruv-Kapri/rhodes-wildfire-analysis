```python
import pystac
items = pystac.ItemCollection.from_file("../outputs/rhodes_sentinel-2.json")
```

```python
assets = items[-1].assets  # last item's asset dictionary
print(assets.keys())
```

```python
for key, asset in assets.items():
    print(f"{key}: {asset.title}")
```

```python
print(assets["thumbnail"].href)
```

```python
print(items[0].assets["thumbnail"].href)
```

```python
import rioxarray
red_href = assets["red"].href
red = rioxarray.open_rasterio(red_href)
print(red)
```

```python
# save whole image to disk
red.rio.to_raster("../outputs/red.tif")
```

```python
red_subset = red.rio.clip_box(
    minx=560900,
    miny=3995000,
    maxx=570900,
    maxy=4015000
)

red_subset.rio.to_raster("../outputs/red_subset.tif")
```
